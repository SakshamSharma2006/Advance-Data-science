```python
from cryptography.fernet import Fernet
from datetime import datetime

# ----------------------------
# KEY GENERATION
# ----------------------------

key = Fernet.generate_key()
cipher = Fernet(key)

# ----------------------------
# CLIENT
# ----------------------------

class Client:

    def send_message(self, message):

        encrypted_message = cipher.encrypt(
            message.encode()
        )

        print("\n===== CLIENT SIDE =====")
        print("Original Message:")
        print(message)

        print("\nEncrypted Message:")
        print(encrypted_message.decode())

        return encrypted_message

# ----------------------------
# SERVER
# ----------------------------

class Server:

    def receive_message(self, encrypted_message):

        decrypted_message = cipher.decrypt(
            encrypted_message
        ).decode()

        print("\n===== SERVER SIDE =====")
        print("Encrypted Received:")
        print(encrypted_message.decode())

        print("\nDecrypted Message:")
        print(decrypted_message)

        now = datetime.now()

        print("\nTime:",
              now.strftime("%H:%M:%S"))

        print("Date:",
              now.strftime("%d-%m-%Y"))

# ----------------------------
# COMMUNICATION
# ----------------------------

client = Client()
server = Server()

message = input("Enter Message: ")

encrypted_data = client.send_message(message)

server.receive_message(encrypted_data)
```
